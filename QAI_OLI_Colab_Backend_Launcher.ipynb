```
# Step 3: Start FastAPI server in background
import subprocess, time, threading

def run_server():
    subprocess.run([
        'uvicorn', 'server:app',
        '--host', '0.0.0.0',
        '--port', '8000'
    ])

t = threading.Thread(target=run_server, daemon=True)
t.start()
print('Backend starting... (models downloading from HF, wait 2-3 min)')
time.sleep(10)
print('Server started!')
```

```
# Step 4: Create public tunnel URL via ngrok
# PASTE YOUR NGROK TOKEN HERE (free at https://dashboard.ngrok.com/get-started/your-authtoken)
NGROK_TOKEN = "PASTE_YOUR_NGROK_TOKEN_HERE"

from pyngrok import ngrok, conf
conf.get_default().auth_token = NGROK_TOKEN

public_url = ngrok.connect(8000)
print('\n' + '='*60)
print('✅ QAI-OLI BACKEND IS LIVE!')
print(f'🌐 Public URL: {public_url}')
print('='*60)
print('\n👉 Copy this URL and set it as BACKEND_URL in Vercel:')
print(f'   {public_url}')
print('\n⚠️  Keep this Colab tab open — closing it will stop the backend.')
```

```
# Step 5: Health check — verify models loaded
import requests, time
time.sleep(5)
try:
    r = requests.get('http://localhost:8000/health')
    print(r.json())
except Exception as e:
    print('Still loading...', e)
```
